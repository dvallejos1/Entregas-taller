```r
# ==========================================================
# TIG 2026: OPTIMIZACIÓN Y MÉTRICAS FINALES (1.000 PORTAFOLIOS)
# ==========================================================

# 1. Librerías
if (!require(tidyquant)) install.packages("tidyquant")
library(tidyquant)
library(plotly)
library(timetk)
library(tidyr)
library(ggplot2)
library(dplyr)
library(PerformanceAnalytics)

# 2. Configuración de Parámetros (Pauta TIG)
# Usamos tus 10 acciones del Excel
tick <- c('MCD', 'MSFT', 'INTC', 'NFLX', 'META', 'AMZN', 'KO', 'AAPL', 'XOM', 'WMT')
inicio <- '2021-04-01'
fin <- '2026-03-31'
rf <- 0.04364  # Tasa Bono USA 10 años (ajustar si tienes una más reciente)

# 3. Descarga de Datos: Acciones y Benchmark (S&P500)
price_data <- tq_get(tick, from = inicio, to = fin, get = 'stock.prices')
benchmark_data <- tq_get("^GSPC", from = inicio, to = fin, get = 'stock.prices')

# 4. Cálculo de Retornos Logarítmicos
log_ret_xts <- price_data %>%
  group_by(symbol) %>%
  tq_transmute(select = adjusted, mutate_fun = periodReturn, period = 'daily', col_rename = 'ret', type = 'log') %>%
  pivot_wider(names_from = symbol, values_from = ret) %>%
  tk_xts(date_var = date)

bench_ret_xts <- benchmark_data %>%
  tq_transmute(select = adjusted, mutate_fun = periodReturn, period = 'daily', col_rename = 'ret', type = 'log') %>%
  tk_xts(date_var = date)

# 5. Matrices Base
mean_ret <- colMeans(log_ret_xts, na.rm = TRUE)
cov_mat <- cov(log_ret_xts, use = "complete.obs") * 252

# 6. Simulación de 1.000 Portafolios (Punto 4 R)
num_port <- 1000
all_wts <- matrix(nrow = num_port, ncol = length(tick))
port_returns <- vector('numeric', length = num_port)
port_risk <- vector('numeric', length = num_port)
sharpe_ratio <- vector('numeric', length = num_port)

set.seed(123)
for (i in 1:num_port) {
  wts <- runif(length(tick))
  wts <- wts / sum(wts)
  all_wts[i, ] <- wts

  # Retorno y Riesgo Anualizado
  port_ret <- sum(wts * mean_ret)
  port_ret <- ((port_ret + 1)^252) - 1
  port_returns[i] <- port_ret

  port_sd <- sqrt(t(wts) %*% (cov_mat %*% wts))
  port_risk[i] <- port_sd

  # Sharpe Ratio
  sharpe_ratio[i] <- (port_ret - rf) / port_sd
}

# Consolidar Resultados
portfolio_values <- tibble(Return = port_returns, Risk = port_risk, SharpeRatio = sharpe_ratio)
colnames(all_wts) <- tick
portfolio_values <- cbind(all_wts, portfolio_values)

# 7. Identificar Portafolios Clave
min_var <- portfolio_values[which.min(portfolio_values$Risk), ]
max_sr <- portfolio_values[which.max(portfolio_values$SharpeRatio), ]

# 8. CÁLCULO DE INDICADORES ADICIONALES (Puntos 4.h al 4.l)
# Usamos el portafolio de Máximo Sharpe como nuestra "mejor cartera"
pesos_optimos <- as.numeric(max_sr[1:length(tick)])
ret_port_optimo <- Return.portfolio(log_ret_xts, weights = pesos_optimos)

beta_port <- CAPM.beta(ret_port_optimo, bench_ret_xts, Rf = rf/252)
treynor_port <- (max_sr$Return - rf) / beta_port
alfa_jensen <- CAPM.jensenAlpha(ret_port_optimo, bench_ret_xts, Rf = rf/252)
track_error <- TrackingError(ret_port_optimo, bench_ret_xts)
info_ratio <- InformationRatio(ret_port_optimo, bench_ret_xts)
m2 <- Modigliani(ret_port_optimo, bench_ret_xts, Rf = rf/252)

# 9. RESULTADOS FINALES
print("--- INDICADORES DE PERFORMANCE (MÁXIMO SHARPE) ---")
print(paste("Sharpe Ratio:", round(max_sr$SharpeRatio, 4)))
print(paste("Treynor Ratio:", round(treynor_port, 4)))
print(paste("Alfa de Jensen:", round(alfa_jensen, 4)))
print(paste("Tracking Error:", round(track_error, 4)))
print(paste("M-Cuadrado (M2):", round(m2, 4)))
print(paste("Ratio de Información:", round(info_ratio, 4)))
print(paste("retornos:", (portfolio_values$Return)))
# 10. Gráfico Final: Frontera + CML
cml_slope <- (max_sr$Return - rf) / max_sr$Risk
ggplot(portfolio_values, aes(x = Risk, y = Return, color = SharpeRatio)) +
  geom_point(alpha = 0.4) +
  theme_minimal() +
  scale_y_continuous(labels = scales::percent) +
  scale_x_continuous(labels = scales::percent) +
  geom_abline(intercept = rf, slope = cml_slope, color = "blue", linetype = "dashed") +
  geom_point(data = max_sr, color = 'red', size = 3) +
  labs(title = "TIG 2026: Frontera Eficiente y CML", x = "Riesgo Anualizado", y = "Retorno Anualizado")
```
